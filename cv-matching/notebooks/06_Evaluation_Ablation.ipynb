```python
llm = pd.read_csv("results/ablation_llm.csv")
manual = [c for c in ["manual_evidence_ok", "manual_invented", "manual_utility_1_3"] if llm[c].notna().any()]
agg = {"n": ("cv_id", "count"), "unsupported_auto": ("unsupported_claims_auto", "mean"), "latence_med_s": ("latency_s", "median")}
agg.update({c: (c, "mean") for c in manual})
llm.groupby("config").agg(**agg).round(3)
```
